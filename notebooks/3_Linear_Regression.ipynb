```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../data/Cleaned_data2022-2025.csv")
```

```python
data.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
```

```python
cat_features = [
    'Driver',
    'Circuit',
    'Year',
    'Compound',
    'TrackStatusSimple',
    'Rainfall',
    'Stint',
    'Position',
]

num_features = [
    'LapNumber',
    'LapsRemaining',
    'TyreLife',
    'AirTemp',
    'Humidity',
    'Pressure',
    'TrackTemp',
    'WindDirection',
    'WindSpeed',
    'LapTimeSeconds',
]

binary_features = [
    'InLap',
    'OutLap'
]
```

### Preprocessing (scaling, onehotencoding)

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("cat",OneHotEncoder(handle_unknown="ignore"),cat_features),
        ("binary","passthrough",binary_features),
        ("num",StandardScaler(),num_features),
    ]
)
```

### Train Test Split

```python
X = data.drop(columns=["TargetLapTime"])
y = data["TargetLapTime"]
```

```python
test_races = [
    "Shanghai",
    "Silverstone",
    "Spa-Francorchamps",
    "Spielberg",
    "Suzuka",
    "São Paulo",
    "Yas Island",
    "Zandvoort"
]
```

```python
test_mask = (
    (data["Circuit"].isin(test_races)) &
    (data["Year"] == 2025)
)
```

```python
train_data = data[~test_mask].copy()
test_data = data[test_mask].copy()
```

```python
X_train = train_data.drop(columns=["TargetLapTime"])
y_train = train_data["TargetLapTime"].copy()

X_test = test_data.drop(columns=["TargetLapTime"])
y_test = test_data["TargetLapTime"].copy()
```

### Model Creation

```python
from sklearn.linear_model import LinearRegression

model_linear_reg = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model_linear_reg.fit(X_train,y_train)
```

```python
predictions_linear_reg = model_linear_reg.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

#Linear
mae_linear_reg = mean_absolute_error(y_test, predictions_linear_reg)
rmse_lienar_reg = root_mean_squared_error(y_test, predictions_linear_reg)

print(f"MAE: {mae_linear_reg:.2f}s")
print(f"RMSE: {rmse_lienar_reg:.2f}s")
```

```python
results = test_data.copy()

# Linear
results["PredictionLinear"] = predictions_linear_reg
results["ErrorLinear"] = results["PredictionLinear"] - results["TargetLapTime"]
results["AbsoluteErrorLinear"] = results["ErrorLinear"].abs()

results
```

```python
print(results["Circuit"].unique())
print(results["Year"].unique())
```

```python
def plot_race(Year, Circuit, Driver):
    race = results[
        (results["Year"] == Year) &
        (results["Circuit"] == Circuit) & 
        (results["Driver"] == Driver)
    ]

    plt.figure(figsize=(12,6))
    
    plt.plot(
        race["LapNumber"],
        race["TargetLapTime"],
        label=f"{Driver} actual",
    )

    plt.plot(
        race["LapNumber"],
        race["PredictionLinear"],
        label=f"{Driver} predicted",
        
    )

    plt.xlabel("Lap Number")
    plt.ylabel("Lap Time (s)")
    plt.title(f"{Driver} {Circuit} {Year} — Actual vs Predicted")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
plot_race(2025,"Silverstone","NOR")
```
